```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn import svm
from sklearn.datasets import make_blobs
import pandas as pd

X_linear, y_linear = make_blobs(n_samples=100, centers=2, random_state=17)

def make_meshgrid(x, y, h=0.02):
    x_min, x_max = x.min() - 1, x.max() + 1
    y_min, y_max = y.min() - 1, y.max() + 1
    xx, yy = np.meshgrid(np.arange(x_min, x_max, h), np.arange(y_min, y_max, h))
    return xx, yy

def plot_contours(ax, clf, xx, yy, **params):
    Z = clf.predict(np.c_[xx.ravel(), yy.ravel()])
    Z = Z.reshape(xx.shape)
    out = ax.contourf(xx, yy, Z, **params)
    return out

kernels = ['linear', 'poly', 'rbf', 'sigmoid']

fig, axes = plt.subplots(2, 2, figsize=(12, 10))

results = []

for kernel, ax in zip(kernels, axes.flatten()):

    clf = svm.SVC(kernel=kernel, gamma='auto')
    clf.fit(X_linear, y_linear)

    accuracy = clf.score(X_linear, y_linear)
    results.append({'Kernel': kernel, 'Accuracy': accuracy})

    X0, X1 = X_linear[:, 0], X_linear[:, 1]
    xx, yy = make_meshgrid(X0, X1)

    plot_contours(ax, clf, xx, yy, cmap=plt.cm.coolwarm, alpha=0.8)
    ax.scatter(X0, X1, c=y_linear, cmap=plt.cm.coolwarm, s=20, edgecolors='k')
    ax.set_xticks(())
    ax.set_yticks(())
    ax.set_title(f'Kernel: {kernel}')

plt.show()

accuracy_df = pd.DataFrame(results)
print("Accuracy Table:")
print(accuracy_df)
```

```python
import pandas as pd
import numpy as np
from sklearn import svm
from sklearn.preprocessing import StandardScaler, LabelEncoder


try:
    df = pd.read_csv('/content/loan_approval_dataset.csv')
except FileNotFoundError:
    print("Error: 'loan_approval_dataset.csv' not found.")
    print("Please make sure the dataset file is in the correct directory.")
    exit()

# 2. Preprocess the Data
# The column names in the CSV have extra spaces, which we need to remove.
df.columns = df.columns.str.strip()

# Convert categorical text columns into numerical format for the model.
le = LabelEncoder()
df['education'] = le.fit_transform(df['education'])
df['self_employed'] = le.fit_transform(df['self_employed'])

# The target variable 'loan_status' also needs to be converted.
df['loan_status'] = le.fit_transform(df['loan_status'])

# 3. Select and Prepare Features
# Let's use 'income_annum' and 'loan_amount' for this analysis.
X = df[['income_annum', 'loan_amount']].values
y = df['loan_status'].values

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

kernels = ['linear', 'poly', 'rbf', 'sigmoid']
results = []

# Loop through kernels to train and evaluate
for kernel in kernels:
    # Initialize and train the SVM classifier
    clf = svm.SVC(kernel=kernel, gamma='auto')
    clf.fit(X_scaled, y)

    # Calculate and store accuracy
    accuracy = clf.score(X_scaled, y)
    results.append({'Kernel': kernel, 'Accuracy': f'{accuracy:.2f}'})

# Create a DataFrame for the accuracy table
accuracy_df = pd.DataFrame(results)
print("Accuracy Table:")
print(accuracy_df)
```
